```python
import psycopg2
from psycopg2.extras import execute_values

conn = psycopg2.connect(host="localhost", port=5432, dbname="nyc", user="nyc", password="nyc")
conn.autocommit = True
with conn.cursor() as cur:
    cur.execute("SELECT COUNT(*) FROM public.rides;")
    print("Rows:", cur.fetchone()[0])

    # Parametrisiert (sicher gegen SQL-Injection)
    cur.execute("""
        SELECT vendor_id, AVG(fare_amount)
        FROM public.rides
        WHERE service_type = %s
        GROUP BY vendor_id
        ORDER BY 1;
    """, ("green",))
    print(cur.fetchall())

    # Bulk-Insert Beispiel
    cur.execute("CREATE TABLE IF NOT EXISTS notes(id serial PRIMARY KEY, note text)")
    data = [("hello",), ("world",)]
    execute_values(cur, "INSERT INTO notes(note) VALUES %s", data)
conn.close()
```
